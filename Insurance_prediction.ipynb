```python
selected_features=['age','is_female', 'bmi', 'children', 'is_smoker',
       'region_northwest', 'region_southeast', 'region_southwest',
     'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese']
```

```python
cat_features=['is_female','is_smoker',
       'region_northwest', 'region_southeast', 'region_southwest',
     'bmi_category_Normal', 'bmi_category_Overweight', 'bmi_category_Obese']
```

```python
from scipy.stats import chi2_contingency
```

```python
import pandas as pd
```

```python
alpha=0.05
```

```python
df_cleaned['charges_bins'] = pd.qcut(df_cleaned['charges'], q = 4, labels=False)
```

```python
chi2_result={}
```

```python
for col in cat_features:
    contingency = pd. crosstab(df_cleaned[col], df_cleaned['charges_bins'])
```

```python
decision = 'Reject Null (keep Features) if p_val<alpha:else, Accept Null( Drop Feature)'
```

```python
chi2_result = {}
for col in cat_features:
    # Generate the contingency table for the current categorical feature and charges_bins
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bins'])

    # Perform the Chi-squared test
    chi2_stat, p_val, dof, expected = chi2_contingency(contingency)

    # Store the results in the chi2_result dictionary
    chi2_result[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }
```

```python
chi2_df = pd.DataFrame(chi2_result).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python
X= final_df.drop("charges" ,axis =1)
Y = final_df["charges"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
print(Y_pred[:5])
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mae=mean_absolute_error(Y_test,Y_pred)
rmse = mean_squared_error(Y_test,Y_pred) ** 0.5
r2 = r2_score(Y_test,Y_pred)
print("MAE:",mae)
print("RMSE:" ,rmse)
print("R2 score:",r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor(n_estimators=100,random_state=42)
rf_model=model.fit(X_train ,Y_train)
```
